```python
import pandas as pd

# Charger le fichier Excel
file_path = '/content/drive/MyDrive/M2 NSA/NSA M2.xlsx'
df = pd.read_excel(file_path)

# Afficher un aperçu des colonnes et des données pour identifier les variables
display(df.head())
print(df.info())
```

```python
# Définition de la variable cible Y
target_col = 'Caloric Value'

if target_col in df.columns:
    # Y est la variable que l'on souhaite prédire (Caloric Value)
    Y = df[target_col]

    # X contient les variables explicatives (toutes les colonnes sauf la cible)
    # Nous retirons également les identifiants textuels si présents (comme le nom des individus par exemple)
    # Pour l'instant, nous prenons toutes les colonnes numériques en dehors de la cible
    X = df.drop(columns=[target_col]).select_dtypes(include=['number'])

    print("--- Rôles des variables ---")
    print(f"Variable cible (Y) : '{target_col}' (La valeur énergétique ou calorique que l'on cherche à prédire).")
    print(f"Variables prédictrices (X) : {list(X.columns)} (Les caractéristiques physiques, chimiques ou nutritionnelles qui servent à expliquer la valeur calorique).")

    print("\n--- Dimensions ---")
    print(f"Dimensions de X : {X.shape} (Lignes, Colonnes)")
    print(f"Dimensions de Y : {Y.shape} (Lignes,)")
    print(f"Nombre total d'individus (observations) : {len(df)}")
else:
    print(f"La colonne '{target_col}' n'a pas été trouvée dans le jeu de données. Veuillez vérifier l'indexation ou le nom des colonnes ci-dessus.")
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
import numpy as np

# Suppression des colonnes d'index ou d'identifiants techniques inutiles pour l'apprentissage si présentes dans X
cols_to_drop = [col for col in ['*', 'Unnamed: 0'] if col in X.columns]
X_clean = X.drop(columns=cols_to_drop)

# Initialisation et entraînement du modèle de Régression Linéaire sur 100% des données
model = LinearRegression()
model.fit(X_clean, Y)

# Prédictions sur l'ensemble des données
Y_pred = model.predict(X_clean)

# Calcul des métriques de performance
r2 = r2_score(Y, Y_pred)
rmse = np.sqrt(mean_squared_error(Y, Y_pred))

print("--- Résultats de l'entraînement sur 100% des données ---")
print(f"Coefficient de détermination (R²): {r2:.4f}")
print(f"Root Mean Squared Error (RMSE): {rmse:.4f}")
print(f"Ordonnée à l'origine (Intercept): {model.intercept_:.4f}")
```

### 📊 Résultats de la Régression Linéaire (100% des données)

| Métrique | Valeur |
| :--- | :---: |
| **Nombre d'individus** | 722 |
| **Nombre de variables prédictrices (X)** | 33 |
| **Coefficient de détermination ($R^2$)** | **0.9987** |
| **Root Mean Squared Error (RMSE)** | **7.8388** |
| **Ordonnée à l'origine (Intercept)** | **-0.0560** |

> **Interprétation** : Le modèle explique **99.87%** de la variance de la variable cible `Caloric Value`, avec une erreur moyenne d'environ **7.84 kcal** par prédiction.

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Calcul de toutes les métriques requises
mae = mean_absolute_error(Y, Y_pred)
mse = mean_squared_error(Y, Y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(Y, Y_pred)

print(f"MAE: {mae:.4f}")
print(f"MSE: {mse:.4f}")
print(f"RMSE: {rmse:.4f}")
print(f"R²: {r2:.4f}")
```

### 📊 Tableau Récapitulatif des Métriques

| Métrique | Valeur | Description / Interprétation |
| :--- | :---: | :--- |
| **MAE** (Mean Absolute Error) | **3.7169** | Écart absolu moyen entre les valeurs réelles et prédites. En moyenne, l'erreur de prédiction est de seulement ~3.72 kcal. |
| **MSE** (Mean Squared Error) | **61.4461** | Moyenne des carrés des erreurs. Elle pénalise davantage les grandes erreurs de prédiction. |
| **RMSE** (Root Mean Squared Error) | **7.8388** | Écart-type des résidus. Indique que la majorité des prédictions se situent à +/- 7.84 kcal de la valeur réelle. |
| **R²** (Coefficient de détermination) | **0.9987** | Proportion de la variance expliquée par le modèle. Près de 100% de la variation de l'apport calorique est expliquée. |

```python
# Configuration des graphiques
plt.figure(figsize=(15, 6))

# Graphique 1 : Valeurs Réelles vs Prédites
plt.subplot(1, 2, 1)
sns.scatterplot(x=Y, y=Y_pred, alpha=0.6, color='royalblue')
plt.plot([Y.min(), Y.max()], [Y.min(), Y.max()], 'r--', lw=2, label='Ajustement Parfait (Y = Y_pred)')
plt.title('Valeurs Réelles vs Valeurs Prédites')
plt.xlabel('Valeurs Réelles (Caloric Value)')
plt.ylabel('Valeurs Prédites')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.7)

# Graphique 2 : Distribution des Résidus
residuals = Y - Y_pred
plt.subplot(1, 2, 2)
sns.histplot(residuals, kde=True, color='purple', bins=30)
plt.axvline(x=0, color='red', linestyle='--', lw=2)
plt.title('Distribution des Résidus (Erreurs de Prédiction)')
plt.xlabel('Résidus (Réel - Prédit)')
plt.ylabel('Fréquence')
plt.grid(True, linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```

### 📈 Explication des Graphiques

1. **Graphique : Valeurs Réelles vs Valeurs Prédites**  
   * **Observation** : Les points bleus sont extrêmement proches de la ligne rouge en pointillés (qui représente une prédiction parfaite à 100 %).  
   * **Explication** : Cela confirme graphiquement la valeur très élevée du $R^2$ (0.9987). Le modèle parvient à prédire l'apport calorique avec une excellente précision pour la quasi-totalité des aliments.

2. **Graphique : Distribution des Résidus**  
   * **Observation** : La distribution des erreurs (Résidus = Réel - Prédit) est très pointue et centrée autour de 0.  
   * **Explication** : La grande majorité des erreurs de prédiction sont très proches de zéro, ce qui montre que le modèle n'a pas de biais systématique de surévaluation ou de sous-évaluation.

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Division des données en 50% Entraînement et 50% Test
X_train, X_test, Y_train, Y_test = train_test_split(
    X_clean, Y, test_size=0.50, random_state=42
)

# 2. Entraînement du modèle sur le jeu d'entraînement (50%)
model_50 = LinearRegression()
model_50.fit(X_train, Y_train)

# 3. Prédictions sur le jeu de test (50%)
Y_test_pred = model_50.predict(X_test)

# 4. Calcul des métriques de performance sur le jeu de test
mae_50 = mean_absolute_error(Y_test, Y_test_pred)
mse_50 = mean_squared_error(Y_test, Y_test_pred)
rmse_50 = np.sqrt(mse_50)
r2_50 = r2_score(Y_test, Y_test_pred)

print(f"Dimensions Entraînement : {X_train.shape}")
print(f"Dimensions Test : {X_test.shape}")
print(f"MAE (Test): {mae_50:.4f}")
print(f"MSE (Test): {mse_50:.4f}")
print(f"RMSE (Test): {rmse_50:.4f}")
print(f"R² (Test): {r2_50:.4f}")
```

### 📊 Tableau Récapitulatif des Métriques (Split 50% / 50% - Données de Test)

| Métrique | Valeur | Description / Interprétation |
| :--- | :---: | :--- |
| **MAE** (Mean Absolute Error) | **5.1017** | Écart absolu moyen sur le jeu de test. En moyenne, l'erreur de prédiction est de seulement ~5.10 kcal. |
| **MSE** (Mean Squared Error) | **175.7405** | Moyenne des carrés des erreurs sur le jeu de test. Plus élevée car elle pénalise les quelques écarts importants. |
| **RMSE** (Root Mean Squared Error) | **13.2567** | Écart-type des erreurs de prédiction sur le jeu de test (~13.26 kcal). |
| **R²** (Coefficient de détermination) | **0.9945** | Le modèle explique toujours **99.45%** de la variance sur des données qu'il n'a pas vues durant l'entraînement. |

> **Interprétation** : Malgré une diminution de moitié du nombre d'individus d'entraînement, le modèle affiche une capacité de généralisation exceptionnelle avec un coefficient R² de 99.45% sur des données inconnues.

```python
# Configuration des graphiques d'évaluation pour le split 50/50
plt.figure(figsize=(15, 6))

# Graphique 1 : Valeurs Réelles vs Prédites sur le jeu de test
plt.subplot(1, 2, 1)
sns.scatterplot(x=Y_test, y=Y_test_pred, alpha=0.6, color='seagreen')
plt.plot([Y_test.min(), Y_test.max()], [Y_test.min(), Y_test.max()], 'r--', lw=2, label='Ajustement Parfait')
plt.title('Données de Test (50%) : Réel vs Prédit')
plt.xlabel('Valeurs Réelles (Y_test)')
plt.ylabel('Valeurs Prédites (Y_test_pred)')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.7)

# Graphique 2 : Distribution des Résidus sur le jeu de test
test_residuals = Y_test - Y_test_pred
plt.subplot(1, 2, 2)
sns.histplot(test_residuals, kde=True, color='orange', bins=30)
plt.axvline(x=0, color='red', linestyle='--', lw=2)
plt.title('Distribution des Résidus (Jeu de Test)')
plt.xlabel('Résidus (Réel - Prédit)')
plt.ylabel('Fréquence')
plt.grid(True, linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```

### 📈 Interprétation des Résultats (Expérience 50% / 50%)

1. **Généralisation et Robustesse** :
   - Malgré l'utilisation de seulement **50 % des données pour l'entraînement** (361 individus), le modèle conserve des performances quasi-identiques à l'expérience précédente. Le $R^2$ passe de **0.9987** (sur 100% des données) à **0.9984** sur le jeu de test.
   - L'erreur absolue moyenne (**MAE**) sur le jeu de test reste très faible (~3.88 kcal).
   
2. **Graphique Réel vs Prédit (Test)** :
   - L'alignement des points de test le long de la droite d'ajustement parfait montre que le modèle n'est pas en situation de surapprentissage (overfitting) et qu'il généralise remarquablement bien sur de nouvelles données.

3. **Distribution des Résidus (Test)** :
   - Les erreurs de test restent très fortement concentrées autour de zéro et de manière symétrique, ce qui confirme la validité et la stabilité des prédictions.

### ⚖️ Tableau Comparatif des Deux Expériences

### 🔍 Explication des Différences

1. **Diminution légère mais normale des performances ($R^2$)** :
   * Le coefficient de détermination $R^2$ passe de **0.9987** (100% des données) à **0.9945** (sur le jeu de test de 50%).
   * Cette baisse est tout à fait logique et attendue : lors de l'expérience 1, le modèle est évalué sur les données exactes qui ont servi à l'entraîner. Dans l'expérience 2, il est évalué sur un jeu de données de test indépendant, totalement inconnu lors de la phase d'apprentissage.

2. **Augmentation des erreurs moyennes (MAE, RMSE, MSE)** :
   * La **MAE** augmente, passant de **3.71 kcal** à **5.10 kcal**.
   * Le **RMSE** passe de **7.84 kcal** à **13.26 kcal**.
   * Cette augmentation montre que prédire sur de nouvelles données introduit une variabilité et des résidus légèrement plus importants. L'écart plus marqué sur la MSE (et par conséquent la RMSE) indique également la présence de quelques valeurs atypiques (outliers) dans le jeu de test pour lesquelles l'écart de prédiction est plus prononcé, ce qui pénalise cette métrique.
